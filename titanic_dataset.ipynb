```python
file_path = '/content/drive/MyDrive/titanic_dataset.csv'

import csv
import time
import numpy as mark

with open(file_path, 'r') as file:
    read = csv.reader(file)
    titanic_list = list(read)

header = titanic_list[0]
content = titanic_list[1:]

titanic_array = mark.array(content)
titanic_tuple = tuple(content)
titanic_set = set(tuple(row) for row in content)
titanic_dict = [dict(zip(header, row)) for row in content]

survivor_array = titanic_array[:, 1]
class_array = titanic_array[:, 2]
sex_array = titanic_array[:, 4]
age_array = titanic_array[:, 5]
fare_array = titanic_array[:, 9]

survivor_int = []
for x in survivor_array:
  if x != "":
    survivor_int.append(int(x))
  else:
    survivor_int.append(0)

class_int = []
for x in class_array:
  if x != "":
    class_int.append(float(x))

sex_identify = []
male_count = 0
female_count = 0
for x in sex_array:
  if x == "male":
    male_count += 1
  elif x == "female":
    female_count += 1

age_float = []
for x in age_array:
  if x != "":
    age_float.append(float(x))

fare_float = []
for x in fare_array:
  if x != "":
    fare_float.append(float(x))
  else:
    fare_float.append(0)

start = time.time()

def survivor_functions():
  survivor_percentage = mark.sum(survivor_int) / len(survivor_int) * 100
  return survivor_percentage

def class_functions():
  class_1 = []
  class_2 = []
  class_3 = []
  for x in class_int:
    if x == 1:
      class_1.append(x)
    elif x == 2:
      class_2.append(x)
    elif x == 3:
      class_3.append(x)
  return class_1, class_2, class_3

def age_functions():
  age_average = (mark.mean(age_float))
  age_median = mark.median(age_float)
  age_min = mark.min(age_float)
  age_max = mark.max(age_float)
  return age_average, age_median, age_min, age_max

def fare_functions():
  max_fare = mark.max(fare_float)
  min_fare = mark.min(fare_float)

  max_ind = mark.argmax(fare_float)
  min_ind = mark.argmin(fare_float)
  if min_fare == 0:
    min_fare = "Free"
  return max_fare, min_fare, max_ind, min_ind

ave, med, min, max = age_functions()
surv_per = survivor_functions()
class_1, class_2, class_3 = class_functions()
max_fare, min_fare, max_ind, min_ind = fare_functions()
max_name = content[max_ind][3]
min_name = content[min_ind][3]

print(f"""----------------------------------------------
First 10 Individuals (List Format):""")
for row in titanic_list[1:10]:
  print(row)


print(f"""----------------------------------------------
Percentage of Survivors: {round(surv_per,2)}%
(Out of {content[-1][0]} Recorded Passengers)
----------------------------------------------""")

print(f"""Age Statistics:
Age Average: {ave}
Age Median: {med}
Age Minimum: {min}
Age Maximum: {max}
----------------------------------------------""")

print(f"""Totals for Each Class:
First Class: {len(class_1)}
Second Class: {len(class_2)}
Third Class: {len(class_3)}
----------------------------------------------""")

print(f"""Total Number of Males and Females:
Male: {male_count}
Female: {female_count}
----------------------------------------------""")

print(f"""Highest and Lowest Fare:
{max_name} -> {max_fare}
{min_name} -> {min_fare}
----------------------------------------------""")

end = time.time()
print(f"Total Time Elapsed: {end - start}")



```
